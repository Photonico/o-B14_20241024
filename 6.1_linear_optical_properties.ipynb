# Linear Optical Properties

```python
# prework

from vmatplot.linear_optical_properties import plot_linear_optical_property
import matplotlib.pyplot as plt

components = [{"xx":"xx-component"}, {"yy":"yy-component"}, {"zz":"zz-component"}]
photon_boundary = (0, 12)
layout = "hor"

# (Thickness of cell, Thickness of system)
factor_bulk = (1, 1)
factor_monolayer = (24.044058323, 1.9870569482029379 + 2*1.91)
factor_bilayer = (28.088116646, 5.59157 + 1.91*2)
factor_bilayer_Hydrogen = (30.088116646, 8.808620 + 1.2*2)

systems = [["o-B₁₄ bilayer with Hydrogen",  "5.1_dielectric_function/bilayer_with_Hydrogen", "orange", None, None, None, factor_bilayer_Hydrogen],
           ["o-B₁₄ bilayer",    "5.1_dielectric_function/bilayer",          "yellow", None, None, None, factor_bilayer],
           ["o-B₁₄ monolayer",  "5.1_dielectric_function/monolayer",        "green", None, None, None, factor_monolayer],
           ["o-B₁₄ bulk",       "5.1_dielectric_function/o-B14_n128_k34",   "blue", None, None, None, factor_bulk]]
```

```python
# Absorption coefficient by GGA-PBE

value_boundary = (-0.0012, 0.03)
plot_linear_optical_property("Absorption coefficient", systems, "absorption", components, layout, True, "eV", photon_boundary, value_boundary, True)
plt.savefig("figures/6.1_absorption.pdf")
```

```python
# Refractive index by GGA-PBE

value_boundary = (-0.3, 7.5)
plot_linear_optical_property("Refractive index", systems, "refractive", components, layout, True, "eV", photon_boundary, value_boundary, True)
plt.savefig("figures/6.2_refractive.pdf")
```

```python
# Extinction coefficient by GGA-PBE

value_boundary = (-0.2, 5)
plot_linear_optical_property("Extinction coefficient", systems, "extinction", components, layout, True, "eV", photon_boundary, value_boundary, True)
plt.savefig("figures/6.3_extinction.pdf")
```

```python
# Reflectivity by GGA-PBE

value_boundary = (-0.04, 1)
plot_linear_optical_property("Reflectivity", systems, "reflectivity", components, layout, True, "eV", photon_boundary, value_boundary, True)
plt.savefig("figures/6.4_reflectivity.pdf")
```

```python
# Energy-loss spectrum by GGA-PBE

value_boundary = (-0.04, 1)
plot_linear_optical_property("Energy-loss spectrum", systems, "energy-loss", components, layout, True, "eV", photon_boundary, value_boundary, True)
plt.savefig("figures/6.5_energy-loss.pdf")
```
