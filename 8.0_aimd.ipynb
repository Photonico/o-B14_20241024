# Ab initio Molecular Dynamics


```python
from vmatplot.aimd import plot_aimd
import matplotlib.pyplot as plt

plot_aimd(
    "AIMD simulation for o-B14 bulk",
    "6.0_AIMD/bilayer_H",
    energy_key="F",
    time_step=1.0,
    x_boundary=(0, 5500),
    temperature_boundary=(0, 800),
    energy_color="Blue",
    temperature_color="Cyan",
    energy_line_weight=1,
    temperature_line_weight=1,
)

plt.savefig("figures/8.0_aimd_bilayer_H.pdf")
fig = plt.gcf()
fig.set_size_inches(10, 6)
plt.savefig("figures_collection/S2.10.pdf")
```
