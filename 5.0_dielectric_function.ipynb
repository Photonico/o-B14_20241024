```python
# Prework

from vmatplot.dielectric_function import *
import matplotlib.pyplot as plt
```

```python
# # Dielectric function

# (Thickness of cell, Thickness of system)
factor_bulk = (1, 1)
factor_monolayer = (24.044058323, 1.9870569482029379 + 2*1.91)
factor_bilayer = (28.088116646, 5.59157 + 1.91*2)
factor_bilayer_Hydrogen = (30.088116646, 8.808620 + 1.2*2)

data_boundary = (-60, 60)

dielectric_list = [["o-B₁₄ bilayer with Hydrogen",  "5.1_dielectric_function/bilayer_with_Hydrogen", "orange",None,None,None, factor_bilayer_Hydrogen],
                   ["o-B₁₄ bilayer",    "5.1_dielectric_function/bilayer",          "yellow",None,None,None,    factor_bilayer],
                   ["o-B₁₄ monolayer",  "5.1_dielectric_function/monolayer",        "green",None,None,None,     factor_monolayer],
                   ["o-B₁₄ bulk",       "5.1_dielectric_function/o-B14_n128_k34",   "blue",None,None,None,      factor_bulk]]
components = [{"xx":"xx-component"},{"yy":"yy-component"},{"zz":"zz-component"}]

plot_dielectric_function("Dielectric function", dielectric_list, components, "hor", True, "eV", (0,12), data_boundary, True)
plt.savefig("figures/5.10_dielectric_o-B14.pdf")

# plot_dielectric_monocomp("Dielectric function", dielectric_list, "yy", "hor", True, "eV", 12, 100, True)
```

```python
# Dielectric function for nbands

dielectric_list = [["(32 nbands)", "5.1_dielectric_function/o-B14_n32_k10", "red",      "solid", 3.2, 1.0, factor_bulk],
                   ["(64 nbands)", "5.1_dielectric_function/o-B14_n64_k10", "orange",   "solid", 2.8, 1.0, factor_bulk],
                   ["(128 nbands)", "5.1_dielectric_function/o-B14_n128_k10", "green",  "solid", 2.4, 1.0, factor_bulk],
                   ["(256 nbands)", "5.1_dielectric_function/o-B14_n256_k10", "blue",   "solid", 2.0, 1.0, factor_bulk]]

plot_dielectric_function("Dielectric function versus NBANDS for o-B₁₄", dielectric_list, components, "hor", True, "eV", (1.24, 32))
plt.savefig("figures/5.11_dielectric_o-B14.pdf")
```
