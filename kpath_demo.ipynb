# The demonstration of K-points path


```python
import matplotlib.pyplot as plt
import numpy as np

from mpl_toolkits.mplot3d.art3d import Line3D

class FrontLine(Line3D):
    def do_3d_projection(self, renderer=None): return -1e9

## unserval settings
fig = plt.figure(figsize=(10, 6),dpi=256)
params = {"text.usetex": False, "font.family": "serif", "mathtext.fontset": "cm", "axes.titlesize": 20, "axes.labelsize": 16, "xtick.labelsize": 14, "ytick.labelsize": 14, "legend.fontsize": 12, "figure.facecolor": "w"}
plt.rcParams.update(params)

len_unit = 12
len_x = len_unit
len_y = len_unit * 0.918
len_z = len_unit * 1.425
shift = 4
rise = 9

ax = fig.add_subplot(111, projection="3d")
ax.set_xlim(-2, 2.5*len_unit)
ax.set_ylim(-2, 2.5*len_unit)
ax.set_zlim(-2, 2.5*len_unit)
ax.view_init(elev=15, azim=30)

# axses
ax.quiver(-shift, -shift, -shift, 2.5*len_unit, 0, 0, color="#C82364", arrow_length_ratio=0.04)  # x-axis
ax.quiver(-shift, -shift, -shift, 0, 2.5*len_unit, 0, color="#19A0A0", arrow_length_ratio=0.04)  # y-axis
ax.quiver(-shift, -shift, -shift, 0, 0, 2.5*len_unit, color="#1478E1", arrow_length_ratio=0.04)  # z-axis

ax.text(2.5*len_unit-shift+2, -shift, -shift, "x", color="#C82364", fontsize=16, ha="center", va="center")
ax.text(-shift, 2.5*len_unit-shift+2, -shift, "y", color="#19A0A0", fontsize=16, ha="center", va="center")
ax.text(-shift, -shift, 2.5*len_unit-shift+2, "z", color="#1478E1", fontsize=16, ha="center", va="center")

# Reciprocal lattice vectors
a, b, c = 5.6799960512011207, 6.1471111863193357, 30.088116646
b1 = np.array([2*np.pi/a, 0, 0])
b2 = np.array([0, 2*np.pi/b, 0])
b3 = np.array([0, 0, 2*np.pi/c])

ax.quiver(0, 0, 0+rise, 0, len_y, 0, color="#F03C64", arrow_length_ratio=0.1)                       # Γ → Z
ax.quiver(0, len_y, 0+rise, len_x, 0, 0, color="#F03C64", arrow_length_ratio=0.1)                   # Z → T
ax.quiver(len_x, len_y, 0+rise, 0, -len_y, 0, color="#F03C64", arrow_length_ratio=0.1)              # T → Y
ax.quiver(len_x, 0, 0+rise, -len_x, 0, 0, color="#FAC828", arrow_length_ratio=0.1)                  # Y → Γ

ax.quiver(len_x, 0, 0+rise, 0, 0, len_z, color="#FA8C00", arrow_length_ratio=0.1)                   # Y → S
ax.quiver(len_x, 0, len_z+rise, -len_x, 0, 0, color="#FA8C00", arrow_length_ratio=0.1)              # S → X
ax.quiver(0, 0, len_z+rise, 0, len_y, 0, color="#FA8C00", arrow_length_ratio=0.1)                   # X → U
ax.quiver(0, len_y, len_z+rise, len_x, 0, 0, color="#FA8C00", arrow_length_ratio=0.1)               # U → R

ax.quiver(0, len_y, 0+rise, 0, 0, len_z, color="#AAAFC3", linestyle="--", arrow_length_ratio=0) 
ax.quiver(len_x, len_y, 0+rise, 0, 0, len_z, color="#AAAFC3", linestyle="--", arrow_length_ratio=0) 
ax.quiver(len_x, 0, len_z+rise, 0, len_y, 0, color="#AAAFC3", linestyle="--", arrow_length_ratio=0) 

ax.quiver(0, len_y, len_z+rise, 0, len_y, 0, color="#6E7387", linestyle="--", arrow_length_ratio=0)
ax.quiver(0, 2*len_y, len_z+rise, 2*len_x, 0, 0, color="#6E7387", linestyle="--", arrow_length_ratio=0)
ax.quiver(2*len_x, 2*len_y, len_z+rise, 0, -2*len_y, 0, color="#6E7387", linestyle="--", arrow_length_ratio=0)
ax.quiver(2*len_x, 0, len_z+rise, -len_x, 0, 0, color="#6E7387", linestyle="--", arrow_length_ratio=0)

ax.quiver(0, 2*len_y, len_z+rise, 0, 0, -2*len_z, color="#6E7387", linestyle="--", arrow_length_ratio=0)
ax.quiver(2*len_x, 2*len_y, len_z+rise, 0, 0, -2*len_z, color="#6E7387", linestyle="--", arrow_length_ratio=0)
ax.quiver(2*len_x, 0, len_z+rise, 0, 0, -2*len_z, color="#6E7387", linestyle="--", arrow_length_ratio=0)
ax.quiver(0, 0, -len_z+rise, 0, 0, 2*len_z, color="#6E7387", arrow_length_ratio=0, linestyle=(9, plt.rcParams["lines.dashed_pattern"]))

ax.quiver(0, 0, -len_z+rise, 0, 2*len_y, 0, color="#6E7387", linestyle="--", arrow_length_ratio=0)
ax.quiver(0, 2*len_y, -len_z+rise, 2*len_x, 0, 0, color="#6E7387", linestyle="--", arrow_length_ratio=0)
ax.quiver(2*len_x, 2*len_y, -len_z+rise, 0, -2*len_y, 0, color="#6E7387", linestyle="--", arrow_length_ratio=0)
ax.quiver(2*len_x, 0, -len_z+rise, -2*len_x, 0, 0, color="#6E7387", linestyle="--", arrow_length_ratio=0)

# Label
ax.text(0, 0+0.5, rise-0.5, "Γ", fontsize=10, color="#C82864", ha="left", va="top")
ax.text(0, len_y+0.5, rise-0.5, "Z", fontsize=10, color="#C82864", ha="left", va="top")
ax.text(len_x, len_y+0.5, rise-0.5, "T", fontsize=10, color="#C82864", ha="left", va="top")
ax.text(len_x, 0.5, rise-0.5, "Y", fontsize=10, color="#C82864", ha="left", va="top")

ax.text(len_x, 0.5, rise+len_z-0.5, "S", fontsize=10, color="#EB731E", ha="left", va="top")
ax.text(0, 0+0.5, rise+len_z-0.5, "X", fontsize=10, color="#EB731E", ha="left", va="top")
ax.text(0, len_y+0.5, rise+len_z-0.5, "U", fontsize=10, color="#EB731E", ha="left", va="top")
ax.text(len_x, len_y+0.5, rise+len_z-0.5, "R", fontsize=10, color="#EB731E", ha="left", va="top")

# ax.set_axis_off()
# plt.savefig("kpath.pdf")

# fig = plt.gcf()
# fig.set_size_inches(5, 6)
# plt.savefig("kpath_s.pdf")

ax.set_axis_off()
ax.set_position([0.00, 0.00, 1.00, 1.00])
fig.subplots_adjust(left=0, right=1, bottom=0, top=1)

# try:
#     ax.set_box_aspect((len_x, len_y, len_z), zoom=1.2)
# except TypeError:
#     ax.set_box_aspect((len_x, len_y, len_z))

fig.set_size_inches(4.5, 4)
plt.savefig("kpath_tide.pdf", bbox_inches="tight", pad_inches=0.00)
```
