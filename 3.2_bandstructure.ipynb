# Bandstructure

```python
# Prework

from vmatplot.bandstructure import plot_bandstructure
import matplotlib.pyplot as plt
```

```python
# o-B₁₄

bs_list = ["monocolor", "", "3.1_bandstructure/o-B14_K48", "blue"]

plot_bandstructure("Band structure for bulk o-B₁₄", bs_list, (-4,4), True)
# plt.savefig("figures/3.21_bandstructure_o-B14.pdf")
```

```python
# monolayer o-B₁₄

bs_list = [["monocolor", "bands of GGA-PBE", "3.1_bandstructure/monolayer", "blue"],
           ["monocolor", "bands of HSE06", "3.1_bandstructure/monolayer_HSE", "orange"],
           ["monocolor", "bands of R2SCAN", "3.1_bandstructure/monolayer_R2SCAN", "lime"]]

plot_bandstructure("Band structure for monolayer o-B₁₄", bs_list, (-4,4), True)
# plt.savefig("figures/3.22_bandstructure_monolayer.pdf")
```

```python
# bilayer o-B₁₄

bs_list = ["monocolor", "", "3.1_bandstructure/bilayer", "blue"]

plot_bandstructure("Band structure for bilayer o-B₁₄", bs_list, (-4,4), True)
# plt.savefig("figures/3.23_bandstructure_bilayer.pdf")
```

```python
# bilayer o-B₁₄ with hydrogen termination

bs_list = ["monocolor", "", "3.1_bandstructure/bilayer_with_Hydrogen", "blue"]

plot_bandstructure("Band structure for bilayer o-B₁₄ with hydrogen termination", bs_list, (-4,4), True)
# plt.savefig("figures/3.24_bandstructure_bilayer_with_Hydrogen.pdf")
```

```python
# monolayer o-B₁₄ with sym break

bs_list = [
           ["monocolor", "bands of GGA-PBE", "3.1_bandstructure/monolayer", "blue"],
           ["monocolor", "bands of shifted atoms", "3.1_bandstructure/monolayer_shifting", "orange"],
           ["monocolor", "bands of symmetry off", "3.1_bandstructure/monolayer_sym_off", "lime"]
           ]

plot_bandstructure("Band structure for monolayer o-B₁₄ for symmetry testing", bs_list, (-4,4), True)
plt.savefig("figures/3.25_bandstructure_monolayer_sym_break.pdf")
```

```python
# bilayer o-B₁₄ HSE

bs_list = ["monocolor", "", "3.1_bandstructure/bilayer_HSE", "blue"]

plot_bandstructure("Band structure for bilayer o-B₁₄", bs_list, (-4,4), True)
plt.savefig("figures/3.26_bandstructure_bilayer_HSE.pdf")
```

```python
# bilayer o-B₁₄ HSE pub

bs_list = ["monocolor", "", "3.1_bandstructure/bilayer_HSE", "blue"]

plot_bandstructure("Band structure for bilayer o-B₁₄", bs_list, (-4,3), True)
fig = plt.gcf()
fig.set_size_inches(8, 6)
plt.savefig("figures_collection/fig2.6c.pdf")
```

```python
# bilayer o-B₁₄ HSE+PBE

bs_list = [["monocolor", "bands of GGA-PBE", "3.1_bandstructure/bilayer", "blue"],
           ["monocolor", "bands of HSE06", "3.1_bandstructure/bilayer_HSE", "purple"],]

plot_bandstructure("Band structure for bilayer o-B₁₄", bs_list, (-4,4), True)
# plt.savefig("figures/3.27_bandstructure_bilayer_compare.pdf")
```
