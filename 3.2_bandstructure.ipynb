# Bandstructure

```python
# Prework

from vmatplot.bandstructure import plot_bandstructure
import matplotlib.pyplot as plt
```

```python
# o-B₁₄

bs_list = ["monocolor", "", "3.1_bandstructure/o-B14_K48", "blue"]

plot_bandstructure("Bandstructure for o-B₁₄", bs_list, (-4,4), True)
plt.savefig("figures/3.21_bandstructure_o-B14.pdf")
```

```python
# o-B₁₄ monolayer

bs_list = [["monocolor", "GGA-PBE", "3.1_bandstructure/monolayer", "blue"],
           ["monocolor", "R2SCAN", "3.1_bandstructure/monolayer_R2SCAN", "lime"]]

plot_bandstructure("Bandstructure for o-B₁₄ monolayer", bs_list, (-4,4), True)
plt.savefig("figures/3.22_bandstructure_monolayer.pdf")
```

```python
# o-B₁₄ bilayer

bs_list = ["monocolor", "", "3.1_bandstructure/bilayer", "blue"]

plot_bandstructure("Bandstructure for B₁₄ o-bilayer", bs_list, (-4,4), True)
plt.savefig("figures/3.23_bandstructure_bilayer.pdf")
```

```python
# o-B₁₄ bilayer with Hydrogen

bs_list = ["monocolor", "", "3.1_bandstructure/bilayer_with_Hydrogen", "blue"]

plot_bandstructure("Bandstructure for o-B₁₄ bilayer with Hydrogen", bs_list, (-4,4), True)
plt.savefig("figures/3.24_bandstructure_bilayer_with_Hydrogen.pdf")
```
