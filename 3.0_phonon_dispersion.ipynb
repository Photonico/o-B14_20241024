# Phonon dispersion


```python
# testing area

from vmatplot.phonon import plot_phonons
import matplotlib.pyplot as plt
```

```python
# monolayer

matters = [["monolayer", "3.0_phonon_dispersion/monolayer", "default", "solid", 1.5, 1.0, 0],
           ["monolayer with Hydrogen", "3.0_phonon_dispersion/monolayer_with_Hydrogen", "orange", "solid", 1.5, 1.0, 0]]

plot_phonons("monolayer phonon dispersion", matters, [-5,10], "upper right")
plt.savefig("figures/3.11_phonon_monolayer.pdf")
```

```python
# bilayer

matters = [["bilayer", "3.0_phonon_dispersion/bilayer", "default", "solid", 1.5, 1.0, 0],
           ["bilayer with Hydrogen", "3.0_phonon_dispersion/bilayer_with_Hydrogen", "orange", "solid", 1.5, 1.0, 0]]

plot_phonons("Bilayer phonon dispersion", matters, [-5,10], "upper right")
plt.savefig("figures/3.12_phonon_bilayer.pdf")
```
