# Phonon dispersion


```python
# testing area

from vmatplot.phonon import plot_phonons
import matplotlib.pyplot as plt
```

```python
# Phonon dispersion 3×3

matters = [
           ["bilayer 3×3", "3.0_phonon_dispersion_vasp/bilayer_3",                       "yellow",   "solid",    1.5, 1.0, 0],
           ["monolayer 3×3", "3.0_phonon_dispersion_vasp/monolayer_3",                   "cyan",     "solid",    1.5, 1.0, 0],
           ]

plot_phonons("Phonon dispersion", matters, [-2,8], "upper right")
plt.savefig("figures/3.13_phonon_noHydrogen.pdf")
```

```python
# Phonon dispersion 3×3

matters = [
           ["bilayer with Hydrogen 3×3", "3.0_phonon_dispersion_phononpy/bilayer_top",      "orange",   "solid",    1.5, 1.0, 0],
           ["monolayer with Hydrogen 3×3", "3.0_phonon_dispersion_phononpy/monolayer_top",  "lime",     "solid",    1.5, 1.0, 0],
           ]

plot_phonons("Phonon dispersion", matters, [-2,8], "upper right")
plt.savefig("figures/3.14_phonon_Hydrogen.pdf")
```

```python
# Phonon dispersion 3×3

matters = [["monolayer 3×3", "3.0_phonon_dispersion_vasp/monolayer_3",                      "cyan",     "solid",    1.5, 1.0, 0],
           ["monolayer with Hydrogen 3×3", "3.0_phonon_dispersion_phononpy/monolayer_top",  "lime",     "solid",    1.5, 1.0, 0],
           ["bilayer 3×3", "3.0_phonon_dispersion_vasp/bilayer_3",                          "yellow",   "solid",    1.5, 1.0, 0],
           ["bilayer with Hydrogen 3×3", "3.0_phonon_dispersion_phononpy/bilayer_top",      "orange",   "solid",    1.5, 1.0, 0],
           ]

plot_phonons("Phonon dispersion", matters, [-2,8], "upper right")
# plt.savefig("figures/3.15_phonon.pdf")
```
