```python
# testing area

from vmatplot.phonon import plot_phonons
import matplotlib.pyplot as plt
```

```python
# Phonon dispersion 2×2

matters = [["monolayer 2×2", "3.0_phonon_dispersion_vasp/monolayer_2",                      "cyan",     "solid",    1.5, 1.0, 0],
           ["H-terminated monolayer 2×2", "3.0_phonon_dispersion_vasp/monolayer_H_2",      "lime",     "solid",    1.5, 1.0, 0],
           ["bilayer 2×2", "3.0_phonon_dispersion_vasp/bilayer_2",                          "yellow",   "solid",    1.5, 1.0, 0],
           ["H-terminated bilayer 2×2", "3.0_phonon_dispersion_vasp/bilayer_H_2",          "orange",   "solid",    1.5, 1.0, 0],
           ]

plot_phonons("Phonon dispersion", matters, [-3,4], "upper right")
plt.savefig("figures/3.11_phonon_2.pdf")

plt.savefig("figures_collection/S2.8.pdf")
```

```python
# Phonon dispersion 3×3

matters = [
           ["bilayer 3×3", "3.0_phonon_dispersion_vasp/bilayer_3",                       "yellow",   "solid",    1.5, 1.0, 0],
           ["monolayer 3×3", "3.0_phonon_dispersion_vasp/monolayer_3",                   "cyan",     "solid",    1.5, 1.0, 0],
           ]

plot_phonons("Phonon dispersion", matters, [-2,8], "upper right")
plt.savefig("figures/3.13_phonon_noHydrogen.pdf")
```
