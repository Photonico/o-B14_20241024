# DoS with spin


```python
# prework

from vmatplot.dos import plot_dos_spin
import matplotlib.pyplot as plt

energy_range = [-14, 6]
```

```python
# FM

matter = [["FM", "4.1_PDoS/monolayer_FM_ollie", "spin-up",  "Orange"],
          ["FM", "4.1_PDoS/monolayer_FM_ollie", "spin-down","Cyan"],
          ["FM", "4.1_PDoS/monolayer_FM_ollie", "total",    "Blue"]]

plot_dos_spin("Spin-polarized DOS of FM monolayer o-B₁₄", matter, energy_range, [-8, 14], dos_quantity="Total")

plt.savefig("figures/4.2a_monolayer_AFM.pdf")
fig = plt.gcf()
fig.set_size_inches(12, 6)
plt.savefig("figures_collection/S2.16a.pdf")
```

```python
# AFM

matter = [["AFM", "4.1_PDoS/monolayer_AFM_ollie", "spin-up",  "Orange"],
          ["AFM", "4.1_PDoS/monolayer_AFM_ollie", "spin-down","Cyan"],
          ["AFM", "4.1_PDoS/monolayer_AFM_ollie", "total",    "Blue"]]

plot_dos_spin( "Spin-polarized DOS of AFM monolayer o-B₁₄", matter, energy_range, [-8, 14], dos_quantity="Total")

plt.savefig("figures/4.2b_monolayer_AFM.pdf")
fig = plt.gcf()
fig.set_size_inches(12, 6)
plt.savefig("figures_collection/S2.16b.pdf")
```
