# Bandstructure with spin

```python
# Prework

from vmatplot.bandstructure import plot_bandstructure, plot_bandstructure_spin
import matplotlib.pyplot as plt
```

```python
spin_list = [["monocolor", "spin up", "3.1_bandstructure/monolayer_FM_HSE06",   "orange", "solid"],
             ["monocolor", "spin down", "3.1_bandstructure/monolayer_FM_HSE06",  "wine", "dashed"],]

plot_bandstructure_spin("Bandstructure for o-B14 monolayer", spin_list, "FM", (-4, 3), True)
plt.savefig("figures/3.23_bs_monolayer_.pdf")

fig = plt.gcf()
fig.set_size_inches(8, 6)
plt.savefig("figures_collection/fig2.6a.pdf")
```

```python
spin_list = [["monocolor", "spin up", "3.1_bandstructure/monolayer_AFM_HSE06",   "orange", "solid"],
             ["monocolor", "spin down", "3.1_bandstructure/monolayer_AFM_HSE06",  "wine", "dashed"],]

plot_bandstructure_spin("Bandstructure for o-B14 monolayer", spin_list, "AFM", (-4, 3), True)
plt.savefig("figures/3.23_bs_monolayer_.pdf")

fig = plt.gcf()
fig.set_size_inches(8, 6)
plt.savefig("figures_collection/fig2.6b.pdf")
```
