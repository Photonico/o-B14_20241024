# Combine PDFs horizontally


```python
import os
import fitz
import matplotlib.pyplot as plt

def combine_pdfs_horizontally(left_pdf, right_pdf, output_pdf, gap=12):
    left_doc = fitz.open(left_pdf)
    right_doc = fitz.open(right_pdf)

    left_page = left_doc[0]
    right_page = right_doc[0]

    left_rect = left_page.rect
    right_rect = right_page.rect

    new_width = left_rect.width + gap + right_rect.width
    new_height = max(left_rect.height, right_rect.height)

    output_doc = fitz.open()
    output_page = output_doc.new_page(width=new_width, height=new_height)

    output_page.show_pdf_page(
        fitz.Rect(0, 0, left_rect.width, left_rect.height),
        left_doc,
        0,
    )
    output_page.show_pdf_page(
        fitz.Rect(left_rect.width + gap, 0,
                  left_rect.width + gap + right_rect.width, right_rect.height),
        right_doc,
        0,
    )
    output_doc.save(output_pdf)
    output_doc.close()
    left_doc.close()
    right_doc.close()
```

```python
# Combine two PDFs horizontally

combine_pdfs_horizontally(
    "7.1_superconductivity_bulk.pdf",
    "7.1_superconductivity_bilayer_H.pdf",
    "7.1_superconductivity.pdf",
    gap=10,
)
```
